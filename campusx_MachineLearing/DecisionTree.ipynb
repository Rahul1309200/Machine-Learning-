```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
X=iris.data
Y=iris.target
Y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.metrics import accuracy_score
from sklearn.tree import plot_tree
from sklearn.tree import DecisionTreeClassifier
```

```python
def changing_params(title,**args):
    clf=DecisionTreeClassifier(**args)
    clf.fit(X_train,Y_train)
    Y_pred=clf.predict(X_test)
    val=accuracy_score(Y_test,Y_pred)
    print("accuracy =" ,val)
    plot_tree(clf)


```

```python
changing_params("Max_depth",max_depth=4)
```

```python
changing_params("min_samples_leaf",min_samples_leaf=5)
```

```python
changing_params("entropy criterion",criterion="entropy")
```

```python
changing_params("splitter",splitter="random")
```

```python
changing_params("min_samples_split",min_samples_split=12)
```

```python
changing_params("max_features",max_features=4)
```
