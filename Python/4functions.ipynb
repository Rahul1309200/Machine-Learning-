```python
# ------- *args and *kwards -----------
# used to pass variable length of arguments in the function

# --- *args ----
# allow us to pass a variable number of non keyword arguments to a function

def multiply(a,b):
    return a*b
print(multiply(2,3))

def multiply_n(*args): # i can use any keyword except args
    result=1
    for i in args:
        result*=i
    print(args,end="=") #this is a tuple
    return result
print(multiply_n(2,2,2,4,2))




# ----- *kwargs -------
# Allows us to pass any number of keywpord arguments
# Key word argument means that they contain a key value pair like python dictionary

def country_capital(**kwargs):
    # Now here kwargs will become dictionary
    for i,j in kwargs.items():
        print(i,"its capital is",j)
    print(kwargs)
country_capital(india="Delhi",Srilanka="columbo")


# -------- Point to remeber ----------
# first normal argument ->args ->kwargs
```

```python
# ----------- Benefits of using a function ---------------
# Code Modularity
# Code Readability
# Code Reusability


# ------------- Lambda Function --------------------
# A lambda function is a small anonymous function 
# A lambda function can take any number of arguments,but can only have one expression


# Paramters one or more paramters are suported must be separated by comma no paranthesis and should have a single valid expression
lambda a,b:a+b

# Square function
a=lambda x:x**2
a(2)

# Sum function
a=lambda x,y:x+y
a(3,4)


# ----- Difference between lmbda and normal function ------
# No name
# Lambda functions returns the full functions only
# Lambda fubnction is always one line
# Not reusable
# They are used in HOF(Higher order functions)
# HOf are the functions in which returns. a functiosn or taken functions as an input

def square(x):
    return x**2
def transform(f,l):
    output=[]
    for i in l:
        output.append(f(i))
    return output
l=[1,2,3,4,5]
m=transform(square,l)
print(m)

# Here we can simply use lambda function
m=transform(lambda x:x**2,l)
print(m)





# check if string as a
a=lambda s:'a' in s
a("Rahul")

# odd or even
a=lambda x:'even' if x%2==0 else "odd"
a(5)
```

```python
# ------------------- Map/Filter/reduce --------------------
# Map wants a lambda function and a iterable like list
# map is basically a higher order function

del list
# ---- Map ----
print(list(map(lambda x:x**2,[1,2,3,4,5])))

# it will map all the values

# odd/even laneling of each value of list
print(list(map(lambda x:"even" if x%2==0 else "odd",[1,2,3,4,5])))


# ------ filter -----
l=[1,2,3,4,5]
print(list(filter(lambda x:x>3,l)))

# -------- Reduce --------
import functools
list=[1,2,3,4,5]
print(functools.reduce(lambda x,y:x+y,[1,2,3,4,5]))
# 1+2=3 [3,3,4,5]
# 3+3=6 [6,4,5]
# 6+4=10 [10,5]
# 10+5=15 

# find min
print(functools.reduce(lambda x,y:x if x<y else y,[1,2,3,4,5]))

```
