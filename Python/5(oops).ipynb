```python
# -------------- Banking Application ------------------
# class keyword to make a class and than classname in Pascalcase(MyAtm)

# Two types of classes user defined and built in(list)
class Atm:
    # Constructor is basically a function inside a class
    # for now dont worry about self these are rules i will explain these later on
    # Constructor is a special function that we need not to call the constructor function it is called automatically
    def __init__(self):
        self.pin=""
        self.balance=0
        # print("Constructor called")
        self.Menu()
    def Menu(self):
        user_input=input("""
        Hi,How can i help you
        1.Press 1 to create Pin
        2. Press 2 to change pin
        3. Press 3 to check Balance
        4. Press 4 to withdraw amount
        5. Anything else to exit
        """)
        if user_input=='1':
            self.create_pin()
        elif user_input=='2':
            self.change_pin()
        elif user_input=='3':
            self.check_balance()
        elif user_input=='4':
            self.withdraw_amount()
        else:
            exit()
    
    def create_pin(self):
        user_pin=input("Enter the new Pin")
        self.pin=user_pin
        user_balance=int(input("enter your balance"))
        self.balance=user_balance
        print("Your Pin is created")
        self.Menu()
    
    def change_pin(self):
        user_pin=input("Enter your password")
        if(user_pin==self.pin):
            new_pin=input("Enter your new pin")
            self.pin=new_pin
            print("Your new pin is setted succesfully")
            self.Menu()
        else:
            input("Sorry buddy wrong pin")
            self.Menu()
    
    def check_balance(self):
        user_pin=input("Enter your password")
        if(user_pin==self.pin):
            print("Your balance is ",self.balance)
            self.Menu()
        else:
            input("Sorry buddy wrong pin")
            self.Menu()
    
    def withdraw_amount(self):
        user_pin=input("Enter your password")
        if(user_pin==self.pin):
            withdraw_amount=int(input("Tell me the amount you want to withdraw"))
            self.balance-=withdraw_amount
            print("Balance updated succesfully")
            self.Menu()
        else:
            input("Sorry buddy wrong pin")
            self.Menu()



    



obj=Atm() #it will call the constructor
print(type(obj))
obj.balance
```

```python
# --------------- Fraction DataType ------------------
# print(3/4*1/2) I dont want in decimal
from math import gcd
class Fraction:
    # Parameterized constructor

    def __init__(self,x,y):
        self.numerator=x;
        self.denominator=y;
    def __str__(self):
        return str(self.numerator)+"/"+str(self.denominator)
    def __add__(self,other):
        new_num=self.numerator*other.denominator+other.numerator*self.denominator
        new_den=self.denominator*other.denominator
        common = gcd(new_num, new_den)
        new_num //= common
        new_den //= common

        return str(new_num) + "/" + str(new_den)
    
    def __sub__(self,other):
        new_num=self.numerator*other.denominator-other.numerator*self.denominator
        new_den=self.denominator*other.denominator
        common = gcd(new_num, new_den)
        new_num //= common
        new_den //= common

        return str(new_num) + "/" + str(new_den)
    
    def __mul__(self,other):
        new_num=self.numerator*other.numerator
        new_den=self.denominator*other.denominator
        common = gcd(new_num, new_den)
        new_num //= common
        new_den //= common

        return str(new_num) + "/" + str(new_den)
    
    def __truediv__(self,other):
        new_num=self.numerator*other.denominator
        new_den=other.numerator*self.denominator
        common = gcd(new_num, new_den)
        new_num //= common
        new_den //= common

        return str(new_num) + "/" + str(new_den)
```
