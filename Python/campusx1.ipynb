```python
# Data Types
# 1.Integer
print(8)
print(1e308) #This is till infinite so it can handle till 10 to the power 308

# Decimal/Float
print(8.5)
print(1.7e308) #This is till infinite so it can handle till 10 to the power 308

# Boolean
print(True)
print(False)

#Text/String
print('Hello world') 

#complex Number
print(5+6j)

# List /Basically thearray in c and little difference
print([1,2,3,4,5])

# Tuple
print((1,2,3,4,5))

# Set
print({1,2,2,3,4,5})

# Dictionary
print({"name":"Rahul","Gender":"Male"})
```
