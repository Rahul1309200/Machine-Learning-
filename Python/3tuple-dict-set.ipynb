```python
# ------------ tuple -----------
# tuple in python is similar to list but it is immutable
# ---- characterstick----
# ordered
# immutable
# allow dublicates


# ----creatung tuple ----
t1=()
print(t1)
# create a single item tuple
t2=(2)
print(t2,type(t2)) # you cant create like this you have to add comma
t2=(2,)
print(t2,type(t2))

# homo tuple
t3=(1,2,3,4)
print(t3)

# heterogenous tuple
t4=(1,"Rahul",[1,2,3])
print(t4)

# tuple within a tuple
t5=(1,2,3,(4,5))
print(t5)

# using tyoe conversion
t6=tuple("hello")
print(t6)
```

```python
# ----- Access items from a tuple -----------
t3
# indexing
print(t3[1])
# slicing
print(t3[0:2])
print(t3[-1])
print(t3[::-1])

print(t5)
print(t5[-1][0])
```

```python
# ----------- operations on tuple ---------
# + *

t1=(1,2,3,4)
t2=(1,2,3,4)
print(t1+t2)
print(t1*3)

# membership
print(1 in t1)
print(6 in t1)
print(1 not in t1)

# iteration
for i in t1:
    print(i)
```

```python
# ------------------ Adding items ---------------
# ---- add ---
s= {1,2,3,4}
s.add(5)
print(s)
# ---- update ------
s.update([5,6,7])
print(s)
```

```python
# ------------- Deleting Items ---------------
# ------ del -------
s={1,2,3,4,5}
del s
# print(s) # name 's' is not defined
# we cant do s[0]

# -------- discard ----------
s={1,2,3,4,5}
s.discard(5)
print(s)
s.discard(10) #No error if not present
print(s)

# --------- Remove ---------
s={1,2,3,4,5}
s.remove(5)
print(s)
# s.remove(10) # Remove will throw an error if not present
print(s)

# ------------ pop ----------
s={1,2,3,4,5}
s.pop() # Random
print(s)

# --------- clear ------------
s={1,2,3,4,5}
s.clear()
print(s)
# makes the setempty
```

```python
# Set operations
s1={1,2,3,4,5}
s2={4,5,6,7,8}
# union |
print(s1|s2)
# Intersection (&)
print(s1&s2)
# Difference (-)
print(s1-s2) # s1 those items not present in s2
# Symmetric diiference 
print(s1^s2) #  except the common everything

# Membership
print(1 in s1)
```

```python
# ---------- Set functions ------------
# len/min/max/sum/sorted
s1={1,2,3,4,5}
s2={4,5,6,7,8}
print(len(s1))
print(min(s1))
print(max(s1))
print(sorted(s1))
print(sorted(s1,reverse=True))
print()


# --- union /update ---
print(s1.union(s2))

# It updates s1 wit the union of s1 ands s2
s1.update(s2)
print(s1)
print(s2)


# ---- Intersection/intersection update ---
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.intersection(s2)
print(s1)
print(s2)

# It updates s1 wit the intersection of s1 ands s2
s1.intersection_update(s2)
print(s1)
print(s2)

# Same with differnce and differnce update
# same for symmetric difference and symmteric differece update

```

```python
# -------- isdisjoint/issubset/issuperset -----
s1={1,2,3,4}
s2={3,4,5,6}
print(s1.isdisjoint(s2)) # basically disjoint are the oneswhich has all values different no common

s1={1,2,3}
s2={1,2,3,4,5}
print(s1.issubset(s2))

print(s2.issuperset(s1))
```
