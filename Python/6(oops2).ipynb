```python
# ---------------- co-ordinate Geometry class -------------------------
import math 
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    
    def __str__(self):
        return "<{},{}>".format(self.x,self.y)
    
    def distance(self,other):
        return math.sqrt((other.x-self.x)**2+(other.y-self.y)**2)

    def distance_from_origin(self):
        return self.distance(Point(0,0))
```

```python
class Line:
    # Ax+By+c
    # user will tell a,b,c
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c

    def __str__(self):
        return "{}x+{}y+{}".format(self.a,self.b,self.c)
    
    def point_line_checker(self,point):
        num=self.a*point.x+self.b*point.y+self.c
        if(num==0):
            print("Yes point is on Line")
        else:
            print("Not on line") 
```

```python
obj1=Point(0,0)
print(obj1)
obj2=Point(1,1)
obj1.distance(obj2)
obj2.distance_from_origin()
```

```python
obj1=Point(1,1)
objl1=Line(1,1,-2)
print(objl1)
objl1.point_line_checker(obj1)
```
