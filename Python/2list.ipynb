```python
# ---------- Lists ----------

# --- what are lists ---
# List is the datatype in which you can store multiple things and acts like a dynamic array
list=[1,"Rahul",3.5,[1,2,3]]
print(list)

# ---- list vs array ----
# 1. Array is of fixed size whereas in list it is not
# 2. convienience - Arrays are homogenous you can only store one datatype and lists are heterogenous
# 3. Python lists are slower 
# 4. It occupies extra space

# Basically lists are stores like referential array having the reference address of the item
a=2
print(id(a)) #adress
list=[1,2,3]
print(id(list[0]))
print(id(list[1]))
print(id(list[2]))

# ---- characterstick of list ----
# 1. ordered [1,2,3] !=[2,1,3]
# 2. Mutable
# 3. Heterogenous
# 4. Items can have dublicate
# 5. are dynamic
# 6. can be nested (List withina list)
# 7. Items can be extracted
# 8. can contain any kind of object

# ---- how to create lists ----
# empty
list=[]
# From function
print(list("Hello"))
print(list)
# 1d list
list=[1,2,3,4,5]
print(list)
# 2d list
list=[[1,2,3],[4,5,6]]
print(list)
# 3d list
list=[[[1,2],[3,4]],[[5,6],[7,8]]]
print(list)

# ---- access items from the list ----
# ---- edit items in the list ----
# ---- deleteing items from the list ----
# ---- operations on list ----
# ---- functions on list ----

```

```python
# --------- Accesing items from the list ----------
# ---- Indexing ----
list=[1,2,3,4,5]
list=[1,2,3,[4,5]]
# positive indexing
print(list[0])
print(list[3][0])
# negative indexing
print(list[-1])
print(list[-3])
print(list[-1][-2])


# ---- slicing ----
list=[1,2,3,4,5]
print(list[:3])
print(list[-3:])
print(list[1:3])
print(list[::-1])
```

```python
# ----------- Adding items in List -------------

# ------ Append ------- to append single item
l=[1,2,3,4,5,6]
l.append(6)
print(l)
# if you will try to append multiple
l.append([6,7,8]) #it will ad it as a single item in the list
print(l)

# ----- extend ----- to append multiple
# Basically append just try to append one and extend try to do it for ,ultipklr breaking and doing
list=[1,2,3]
list.extend([6,7,8])
print(list)
list.extend("Delhi")
print(list)

# --------- Insert ---------- to insert at a particular location
list.insert(1,100)
print(list)


```

```python
# ------- Editing items in the List ----------
# IN string you cat do this 
list=[1,2,3]
list[0]=5
print(list)

list[0:2]=[500,600]
print(list)
```

```python
# -------- Deleting items in List ------
# --- del ----
l=[1,2,3]
del l
# print(l) #name 'l' is not defined
l=[1,2,3,4]
# del l[-1]
del l[0:2]
print(l)

# ---- remove ---- basically remove the value not the index
ls=[1,2,3,4,5]
ls.remove(5)
print(ls)

# ------ pop -------
ls.pop(0) #provide the index
ls.pop() #remove last
print(ls)

# ----- clear -----
# basically it makes the liste empty
ls.clear()
print(ls)
```

```python
# ----------------- operators on List -------------------
# ------ Arithematic --------
l1=[1,2,3,4]
l2=[1,2,3,4]
# concatenation or merging
print(l1+l2)
print(l1*3) #it will repeat three times and merge

print(2 in l1)
print(8 in l1)
print(8 not in l1)
print(2 not in l1)
```

```python
# --------- Function in list ---------
# len()
# min
# max
# sorted
list=[1,2,3,4,2,3]
print(len(list))
print(max(list))
print(min(list))
print(sorted(list)) # it is not a permanent operation
print(sorted(list,reverse=True))
```

```python
# ----- count ----- count frequency
list=[1,2,3,4,2,3]
list.count(2)
# index --- tell the index of character
list.index(2) #1srt occurance
# ----- reverse -----
print(list.reverse()) #permanently reverses the list
print(list)
```

```python
# sort vs sorted
# basically sort do the permanent chnages
list=[4,5,6,1,2]
print(sorted(list))
print(list)
list.sort()
print(list)
```

```python
# -------- copy ----------- creating a shallow copy
l=[2,1,5,7,0]
print(id(l))
l1=l.copy()
print(l1)
print(id(l1))
```

```python
# ---------- List comprehension ------------
# it prvides a concise way of creating list
# newlist=[expression for item in iterable if condition==True]
# more time efficuent and space efficient than loop

# -------- Add 1 to 10 number in the list -------
newlist=[i for i in range(1,11)]
print(newlist)

# -------- scalar multiplication on vector -------
v=[2,3,4]
s=3
newlist=[i*s for i in v]
print(newlist)

# -------- Add squares -------
ls=[2,3,4,5,6]
newlist=[i*i for i in ls]
print(newlist)

# ------- print all the numbers divisible by 5 from 1 to 50 -------
newlist=[i for i in range(1,51) if i%5==0]
print(newlist)

# ------- find languages which start with letter p -------
languages=["java","python","php","cpp","html"]
newlist=[language for language in languages if language.startswith("p")]
print(newlist)


# ---------------- Nested if with list comprehension -----------------
# -----i have to create a list if it is in basket 1 and starts with a
basket1=["apple","banana","litchi","apple"]
basket2=["apple","banana","litchi","appe","orange"]
newlist=[fruit for fruit in basket2 if fruit in basket1 if fruit.startswith("a")]
print(newlist)


# --------- Nested list comprehnesion --------
[[i*j for i in range(1,4)] for j in range(1,4)]

# cartesian product
l1=[1,2,3,4]
l2=[5,6,7,8]
newlist=[i*j for i in l1 for j in l2]
```
