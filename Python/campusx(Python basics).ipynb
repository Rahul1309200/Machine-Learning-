```python

def subtractProductAndSum(n: int) -> int:
        product=1
        sum=0
        while n>0:
            rem=n%10
            sum+=rem
            product*=rem
            n=n//10
        return product-sum

print(subtractProductAndSum(241))
```

```python
def numberOfSteps(num: int) -> int:
        if num==0:
            return 0
        counter=0
        while num!=0:
            if num%2==0:
                num=num//2
            else:
                num-=1
            counter+=1
        return counter
print(numberOfSteps(14))
```

```python
# Pattern2 
# 1
# 1 2 1
# 1 2 3 2 1
# 1 2 3 4 3 2 1 
n=4
for i in range(1,n+1):
    for j in range(0,2*i-1):
        if(i<=j):
            print(2*i-j-1,end=" ")
        else:
            print(j+1,end=" ")
    print()
    
for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end=" ")
    for j in range(i-1,0,-1):
        print(j,end=" ")
    print()
```
