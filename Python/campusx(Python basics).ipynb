```python

def subtractProductAndSum(n: int) -> int:
        product=1
        sum=0
        while n>0:
            rem=n%10
            sum+=rem
            product*=rem
            n=n//10
        return product-sum

print(subtractProductAndSum(241))
```

```python
def numberOfSteps(num: int) -> int:
        if num==0:
            return 0
        counter=0
        while num!=0:
            if num%2==0:
                num=num//2
            else:
                num-=1
            counter+=1
        return counter
print(numberOfSteps(14))
```
